# Section 4: Stacks and Queues

Stacks (LIFO) and queues (FIFO) are essential data structures for many interview problems. In Python, lists serve as stacks and `collections.deque` as efficient queues.

### Challenge 17: Stack Push and Pop

A **stack** is LIFO (Last-In, First-Out). In Python, use a list: `.append()` to push, `.pop()` to pop. Both are **O(1)**.

```python
stack = []
stack.append(1)
stack.append(2)
top = stack.pop()  # 2
```

from typing import List


def reverse_list(lst: List[int]) -> List[int]:
    stack = []
    for item in lst:
        stack.append(item)
    result = []
    while stack:
        result.append(stack.pop())
    return result


def is_balanced(s: str) -> bool:
    stack = []
    matching = {')': '(', ']': '[', '}': '{'}
    for char in s:
        if char in '([{':
            stack.append(char)
        elif char in ')]}':
            if not stack or stack.pop() != matching[char]:
                return False
    return len(stack) == 0

### Challenge 18: Queue Enqueue and Dequeue

A **queue** is FIFO (First-In, First-Out). Use `collections.deque` for **O(1)** enqueue (`.append()`) and dequeue (`.popleft()`).

```python
from collections import deque
q = deque([1, 2, 3])
q.append(4)      # enqueue
front = q.popleft()  # dequeue -> 1
```

from typing import List
from collections import deque


def process_queue(tasks: List[str]) -> List[str]:
    q = deque(tasks)
    result = []
    while q:
        result.append(q.popleft())
    return result


def first_n_elements(lst: List[int], n: int) -> List[int]:
    q = deque(lst)
    result = []
    for _ in range(n):
        result.append(q.popleft())
    return result

### Challenge 19: Double Ended Queue

A `deque` supports efficient operations on **both** ends: `.append()`, `.appendleft()`, `.pop()`, `.popleft()` — all **O(1)**.

```python
from collections import deque
d = deque([1, 2, 3])
d.appendleft(0)  # [0, 1, 2, 3]
d.append(4)      # [0, 1, 2, 3, 4]
d.popleft()      # removes 0
d.pop()          # removes 4
```

from typing import List
from collections import deque


def rotate_left(lst: List[int], k: int) -> List[int]:
    d = deque(lst)
    for _ in range(k):
        d.append(d.popleft())
    return list(d)


def palindrome_check(s: str) -> bool:
    d = deque(s)
    while len(d) > 1:
        if d.popleft() != d.pop():
            return False
    return True


# do not modify below this line
print(rotate_left([1, 2, 3, 4, 5], 2))
