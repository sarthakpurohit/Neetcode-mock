from typing import List
import heapq


def sorted_from_heap(elements: List[int]) -> List[int]:
    heap = elements[:]
    heapq.heapify(heap)
    result = []
    while heap:
        result.append(heapq.heappop(heap))
    return result


def get_k_smallest(elements: List[int], k: int) -> List[int]:
    heap = elements[:]
    heapq.heapify(heap)

    return result

### Challenge 34: Max Heap
