### Challenge 0: Sort Ascending

In Python, you can sort a list of elements by calling `.sort()` on the list.

```python
elements = [5, 3, 6, 2, 1]
elements.sort()
print(elements) # [1, 2, 3, 5, 6]
```

By default, the `.sort()` method sorts the elements in ascending order in-place. The return value of the `.sort()` method is `None`.

This method also works for a list of strings.

```python
elements = ["grape", "apple", "banana", "orange"]
elements.sort()
print(elements) # ['apple', 'banana', 'grape', 'orange']
```

By default, strings are sorted in lexicographical order.

### Challenge 1: Sort Descending

The `.sort()` method also accepts optional parameters:

```python
def sort(key=None, reverse=False) -> None:
```

The `key` parameter customizes sorting order. The `reverse` parameter determines whether to sort in descending order (default `False`).

```python
elements = [5, 3, 6, 2, 1]
elements.sort(reverse=True)
print(elements) # [6, 5, 3, 2, 1]
```

You can also sort ascending and then reverse manually:

```python
elements.sort()
elements.reverse()
```

### Challenge 2: Sort Custom

We can specify a custom sorting order using the `key` parameter, which accepts a function that returns a value used for sorting.

```python
def get_word_length(word: str) -> int:
    return len(word)

words = ["apple", "banana", "kiwi", "pear", "watermelon", "blueberry", "cherry"]
words.sort(key=get_word_length)
print(words) # ['kiwi', 'pear', 'apple', 'banana', 'cherry', 'blueberry', 'watermelon']
```

The words are sorted by their length in ascending order, not lexicographically.

### Challenge 3: Sort Lambda

Lambda functions let you define simple functions inline without a separate `def` statement.

```python
words = ["apple", "banana", "kiwi", "pear", "watermelon", "blueberry", "cherry"]
words.sort(key=lambda word: len(word))
print(words) # ['kiwi', 'pear', 'apple', 'banana', 'cherry', 'blueberry', 'watermelon']
```

The syntax is: `lambda parameter: expression`. A lambda must be a single expression and cannot contain multiple statements. It's a convenient shortcut for simple key functions.

### Challenge 4: Sorted Copy

The `sorted()` function returns a **new** sorted list, leaving the original unchanged.

```python
words = ["kiwi", "pear", "apple", "banana", "cherry", "blueberry"]
sorted_words = sorted(words)
print(sorted_words) # ['apple', 'banana', 'blueberry', 'cherry', 'kiwi', 'pear']
```

It accepts `reverse` and `key` parameters just like `.sort()`:

```python
numbers = [5, -3, 2, -4, 6, -2, 4]
sorted_numbers = sorted(numbers, key=abs)
print(sorted_numbers) # [2, -2, -3, 4, -4, 5, 6]
```
