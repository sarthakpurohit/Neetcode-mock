```python
from sortedcontainers import SortedDict


def create_sorted_phonebook(entries: list) -> dict:
    sd = SortedDict()
    for name, number in entries:
        sd[name] = number
    return sd


def range_query(sd, low: str, high: str) -> list:
    result = []
    for key in sd.irange(low, high):
        result.append((key, sd[key]))
    return result
```

### Challenge 39: Sorted Set Basics

`SortedList` maintains elements in sorted order with **O(log n)** add/remove and binary search via `bisect_left`/`bisect_right`.

```python
from sortedcontainers import SortedList
sl = SortedList([5, 1, 3])
print(sl)  # SortedList([1, 3, 5])
sl.add(2)  # SortedList([1, 2, 3, 5])
```

```python
from sortedcontainers import SortedList


def add_and_find_median(numbers: list) -> float:
    sl = SortedList()
    for num in numbers:
        sl.add(num)
    n = len(sl)
    if n % 2 == 1:
        return sl[n // 2]
    else:
        return (sl[n // 2 - 1] + sl[n // 2]) / 2


def count_in_range(numbers: list, low: int, high: int) -> int:
    sl = SortedList(numbers)
    left = sl.bisect_left(low)
    right = sl.bisect_right(high)
    return right - left

print(add_and_find_median([5, 3, 8, 1, 2]))
print(add_and_find_median([5, 3, 8, 1]))
print(count_in_range([1, 3, 5, 7, 9, 11, 13, 15], 4, 12))
```
