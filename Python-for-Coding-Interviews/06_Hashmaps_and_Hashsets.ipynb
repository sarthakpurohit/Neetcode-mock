```python
def word_lengths(words: List[str]) -> Dict[str, int]:
```

```python
from typing import List, Dict
from collections import defaultdict


def group_by_length(words: List[str]) -> Dict[int, List[str]]:
    groups = defaultdict(list)
    for word in words:
        groups[len(word)].append(word)
    return groups


def group_anagrams(words: List[str]) -> List[List[str]]:
    groups = defaultdict(list)
    for word in words:
        key = tuple(sorted(word))
        groups[key].append(word)
    return list(groups.values())
```

```python
from typing import List


def unique_elements(lst: List[int]) -> List[int]:
    return list(set(lst))


def has_duplicates(lst: List[int]) -> bool:

def common_elements(lst1: List[int], lst2: List[int]) -> List[int]:
    return list(set(lst1) & set(lst2))
```

```python
from typing import Dict, Tuple, List


def create_grid_map(rows: int, cols: int) -> Dict[Tuple[int, int], int]:
    result = {}
    counter = 0
    for r in range(rows):
        for c in range(cols):
            result[(r, c)] = counter
            counter += 1
    return result


def count_points(points: List[Tuple[int, int]]) -> Dict[Tuple[int, int], int]:
    result = {}
    for point in points:
        if point in result:
            result[point] += 1
        else:
            result[point] = 1
    return result
```
