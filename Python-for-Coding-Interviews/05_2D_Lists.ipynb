### Challenge 20: Multi-dimensional List

A 2-D list is a list of lists. Access elements with `matrix[row][col]`.

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[1][2])  # 6
```

**Never** use `[[0]*cols]*rows` — all rows share the same reference. Use a comprehension: `[[0 for _ in range(cols)] for _ in range(rows)]`.

def create_matrix(rows: int, cols: int, value: int) -> List[List[int]]:

from typing import List


def sum_row(grid: List[List[int]], row: int) -> int:
    return sum(grid[row])


def sum_column(grid: List[List[int]], col: int) -> int:
    total = 0
    for row in grid:
        total += row[col]
    return total


def get_neighbors(grid: List[List[int]], row: int, col: int) -> List[int]:
    neighbors = []
    rows, cols = len(grid), len(grid[0])
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = row + dr, col + dc
        if 0 <= nr < rows and 0 <= nc < cols:
            neighbors.append(grid[nr][nc])
    return neighbors

from typing import List


def identity_matrix(n: int) -> List[List[int]]:
    return [[1 if i == j else 0 for j in range(n)] for i in range(n)]

    return [[(i + 1) * (j + 1) for j in range(n)] for i in range(n)]
