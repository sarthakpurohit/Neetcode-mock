### Challenge 40: Two Pointers Pattern

**Two pointers** use two indices moving toward each other or in the same direction, often reducing O(n²) to O(n).

```python
def two_sum(nums, target):
    left, right = 0, len(nums) - 1
    while left < right:
        s = nums[left] + nums[right]
        if s == target: return (left, right)
        elif s < target: left += 1
        else: right -= 1
    return (-1, -1)
```

def two_sum_sorted(numbers: List[int], target: int) -> Tuple[int, int]:
    left, right = 0, len(numbers) - 1
    while left < right:
        total = numbers[left] + numbers[right]
        if total == target:
            return (left, right)
        elif total < target:
            left += 1
        else:
            right -= 1
    return (-1, -1)


def is_palindrome(s: str) -> bool:
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True


def remove_duplicates(nums: List[int]) -> int:
    if not nums:
        return 0
    write = 1
    for read in range(1, len(nums)):
        if nums[read] != nums[write - 1]:
            nums[write] = nums[read]
            write += 1
    return write

### Challenge 41: Sliding Window Pattern

The **sliding window** efficiently processes contiguous subarrays by adding/removing one element at a time.

**Fixed window:** slide a window of size `k` across the array.
**Variable window:** expand `right`, shrink `left` when a condition breaks.

```python
def max_sum(nums, k):
    window = sum(nums[:k])
    best = window
    for i in range(k, len(nums)):
        window += nums[i] - nums[i - k]
        best = max(best, window)
    return best
```

from typing import List


def max_subarray_sum(numbers: List[int], k: int) -> int:
    window_sum = sum(numbers[:k])
    max_sum = window_sum
    for i in range(k, len(numbers)):
        window_sum += numbers[i] - numbers[i - k]
        max_sum = max(max_sum, window_sum)
    return max_sum


def longest_unique_substring(s: str) -> int:
    seen = set()
    left = 0
    max_len = 0
    for right in range(len(s)):
        while s[right] in seen:
            seen.remove(s[left])
            left += 1
        seen.add(s[right])
        max_len = max(max_len, right - left + 1)
    return max_len

### Challenge 42: Binary Search Pattern

**Binary search** finds a target in a sorted array in **O(log n)** by halving the search space repeatedly.

```python
def binary_search(nums, target):
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target: return mid
        elif nums[mid] < target: lo = mid + 1
        else: hi = mid - 1
    return -1
```

### Challenge 43: Dynamic Programming Intro

**Dynamic Programming** solves problems by storing intermediate results. **Tabulation** (bottom-up) fills a table iteratively.

```python
def fibonacci(n):
    dp = [0] * (n + 1)
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]
```

from typing import List


def fibonacci(n: int) -> int:
    if n <= 1:
        return n
    dp = [0] * (n + 1)
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]


def climb_stairs(n: int) -> int:
    if n <= 2:
        return n
    dp = [0] * (n + 1)
    dp[1] = 1
    dp[2] = 2
    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]


def max_subarray(nums: List[int]) -> int:
    current_sum = nums[0]
    max_sum = nums[0]
    for i in range(1, len(nums)):
        current_sum = max(nums[i], current_sum + nums[i])
        max_sum = max(max_sum, current_sum)
    return max_sum
