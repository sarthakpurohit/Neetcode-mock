### Challenge 5: Unpacking

Python allows **unpacking** — assigning multiple variables from a list or tuple in one line.

```python
point1 = [0, 0]
point2 = [2, 4]
x1, y1 = point1
x2, y2 = point2
slope = (y2 - y1) / (x2 - x1)
print(slope) # 2.0
```

This is equivalent to `x1, y1 = point1[0], point1[1]` but much cleaner. Unpacking works with tuples and sets too. If the number of variables doesn't match, you get a `ValueError`.

### Challenge 6: Loop Unpacking

You can unpack directly inside `for` loops when iterating over lists of tuples or lists.

```python
students = [("Alice", 90), ("Bob", 85), ("Charlie", 92)]
for name, score in students:
    print(f"{name} scored {score}")
```

This replaces the more verbose `student[0]`, `student[1]` indexing and makes code much cleaner.

### Challenge 7: Enumerate

`enumerate()` gives you both the **index** and **value** when iterating over a list.

```python
fruits = ["apple", "banana", "cherry"]
for i, fruit in enumerate(fruits):
    print(f"{i}: {fruit}")
# 0: apple  1: banana  2: cherry
```

This replaces manual index tracking or `range(len(...))` patterns. You can also set a custom start: `enumerate(fruits, start=1)`.

### Challenge 8: Zip

`zip()` iterates over **multiple lists in parallel**, pairing elements by position.

```python
names = ["Alice", "Bob", "Charlie"]
scores = [90, 85, 92]
for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

If lists have different lengths, `zip()` stops at the shortest. You can zip more than two lists together.

### Challenge 10: Min Max Shortcut

`min()` and `max()` work on any iterable and accept an optional `key` parameter.
