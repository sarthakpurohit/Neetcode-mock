### Challenge 12: Resizable List Part 2

Beyond `.append()`, lists offer `insert()`, `pop()`, and `remove()`.


### Challenge 15: List Clone

Assigning a list to a new variable creates an **alias**, not a copy. Use `.copy()`, `[:]`, or `list()` for a shallow copy.

```python

clone = original.copy()
clone.append(4)
print(original)  # [1, 2, 3] — unchanged
```

```python
from typing import List


def safe_append(lst: List[int], value: int) -> List[int]:
    clone = lst.copy()
    clone.append(value)
    return clone


def double_values(lst: List[int]) -> List[int]:
    clone = lst[:]
    for i in range(len(clone)):
        clone[i] *= 2
    return clone
```

### Challenge 16: List Comprehension

```python
from typing import List


def squares(n: int) -> List[int]:
    return [x**2 for x in range(n)]
```
