### Challenge 16: Class Attributes

Assigning in `__init__` creates instance attributes. Assigning in the class body creates class attributes—often used for shared constants or counters.

### Challenge

### Challenge 17: Class vs Instance Attributes

Mutable class attributes are shared: be careful with lists/dicts on the class. Prefer instance attributes for per-object mutable state.

### Challenge

Fix `Team` so each instance has its own `members` list (do not share one list).

````python
class Team:
    members = []

    def __init__(self, name: str) -> None:
        self.name = name
        # BUG: everyone shares this list. Give each instance its own members list.


t1 = Team("A")
t2 = Team("B")
t1.members.append("x")
print(len(t1.members), len(t2.members))
````

<summary><strong>Show Solution</strong></summary>

```python
class Team:
    def __init__(self, name: str) -> None:
        self.name = name
        self.members = []


t1 = Team("A")
t2 = Team("B")
t1.members.append("x")
print(len(t1.members), len(t2.members))
```

</details>

### Challenge 18: Class Methods

`cls` allows alternate constructors, for example `from_tuple`.

### Challenge

Add `Point.from_tuple(t: tuple[int, int]) -> Point` classmethod.

````python
class Point:
    def __init__(self, x: int, y: int) -> None:
        self.x = x
        self.y = y

    # TODO: from_tuple


print(Point.from_tuple((3, 4)).x, Point.from_tuple((3, 4)).y)
````

<details>
<summary><strong>Show Solution</strong></summary>

```python
class Point:
    def __init__(self, x: int, y: int) -> None:
        self.x = x
        self.y = y

    @classmethod
    def from_tuple(cls, t: tuple[int, int]) -> "Point":
        x, y = t
        return cls(x, y)


print(Point.from_tuple((3, 4)).x, Point.from_tuple((3, 4)).y)
```

</details>

### Challenge 19: Static Methods

Use `@staticmethod` when you want namespacing but do not need `self`/`cls`.

### Challenge

Add `MathUtil.clamp(x, lo, hi)` static method.

````python
class MathUtil:
    # TODO: clamp
    pass


print(MathUtil.clamp(10, 0, 5), MathUtil.clamp(-3, -1, 1))
````

<details>
<summary><strong>Show Solution</strong></summary>

```python
class MathUtil:
    @staticmethod
    def clamp(x: float, lo: float, hi: float) -> float:
        return max(lo, min(hi, x))


print(MathUtil.clamp(10, 0, 5), MathUtil.clamp(-3, -1, 1))
```

</details>

### Challenge 20: Implement Power Calculation System

Combine patterns: class state + methods + validation.

### Challenge

Implement `PowerSystem` with class attribute `grid_load` starting at 0. Method `register(hero_power: int)` adds to load; `status()` returns `"ok"` if load <= 100 else `"overload"`.

````python
class PowerSystem:
    # TODO
    pass


PowerSystem.register(30)
PowerSystem.register(80)
print(PowerSystem.status())
````

<details>
<summary><strong>Show Solution</strong></summary>

```python
class PowerSystem:
    grid_load = 0

    @classmethod
    def register(cls, hero_power: int) -> None:
        cls.grid_load += int(hero_power)

    @classmethod
    def status(cls) -> str:
        return "ok" if cls.grid_load <= 100 else "overload"


PowerSystem.register(30)
PowerSystem.register(80)
print(PowerSystem.status())
```

</details>
