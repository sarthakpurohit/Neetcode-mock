from abc import ABC, abstractmethod


class Pricing(ABC):
    @abstractmethod
    def price(self, n: int) -> int:
        raise NotImplementedError


class BulkPricing(Pricing):
    def price(self, n: int) -> int:
        return n * 8 if n >= 10 else n * 10


class Cart:
    def __init__(self, pricing: Pricing) -> None:
        self.pricing = pricing

    def checkout(self, n: int) -> int:
        return self.pricing.price(n)


print(Cart(BulkPricing()).checkout(12))

    def __init__(self, clock) -> None:
        self.clock = clock

    def hi(self) -> str:
        ts = self.clock()
