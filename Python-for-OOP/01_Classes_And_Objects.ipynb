Classes bundle data and behavior into reusable blueprints called **objects**. In interview-style Python you will model problems as small types with clear state and methods. ### Challenge 0: Intro to Classes

Imagine you are building a game with many characters. Without classes you might repeat variables like `hero_1_name`, `hero_2_name`, and so on. A **class** is a blueprint; an **object** is one concrete instance built from that blueprint.

In Python, `class Name:` introduces a type. The special method `__init__` runs when you construct an object: `Pet("Fluffy", "cat")` passes arguments into `__init__` so you can store them on the instance as **attributes** like `self.name`.

The first parameter of instance methods is traditionally named `self`. It refers to the object being created or acted on. Python passes it for you; you should declare it on each instance method including `__init__`.

**Class names** use **PascalCase** (`Pet`, `SuperHero`).

class SuperHero:

        self.power = power
