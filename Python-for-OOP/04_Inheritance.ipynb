<details>
<summary><strong>Show Solution</strong></summary>

```python
class Character:
    def __init__(self, hp: int) -> None:
        self.hp = hp

    def take_damage(self, amount: int) -> None:
        self.hp = max(0, self.hp - amount)


class Warrior(Character):
    def take_damage(self, amount: int) -> None:
        super().take_damage(int(amount * 0.8))


class Mage(Character):
    def strike(self, other: Character, amount: int) -> None:
        other.take_damage(amount + 5)


w = Warrior(100)
m = Mage(50)
m.strike(w, 10)
print(w.hp)
```
