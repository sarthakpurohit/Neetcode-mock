# Section 6: Abstraction
*Challenges 32–36*

**Abstraction** hides complexity behind a simple interface. In Python, `abc.ABC` with `@abstractmethod` enforces that subclasses implement required methods—useful for plugin-like designs and for communicating intent in interviews.

### Challenge 32: Abstraction

Abstract concepts describe *what* operations exist without dictating *how* each subtype implements them.

### Challenge

Create abstract base `Logger` with abstract `log(msg: str) -> None`.

````python
from abc import ABC, abstractmethod


class Logger(ABC):
    # TODO: abstract log
    pass


class PrintLogger(Logger):
    def log(self, msg: str) -> None:
        print(msg)


PrintLogger().log("hi")
````

<details>
<summary><strong>Show Solution</strong></summary>

```python
from abc import ABC, abstractmethod


class Logger(ABC):
    @abstractmethod
    def log(self, msg: str) -> None:
        raise NotImplementedError


class PrintLogger(Logger):
    def log(self, msg: str) -> None:
        print(msg)


PrintLogger().log("hi")
```

</details>

### Challenge 34: Interface

Python’s structural typing (Protocols) is modern interface-like typing; ABCs are explicit interface contracts.

### Challenge

Implement `Serializable` ABC requiring `to_json() -> str`.

````python
import json
from abc import ABC, abstractmethod


class Serializable(ABC):
    # TODO
    pass


class Point(Serializable):
    def __init__(self, x: int, y: int) -> None:
        self.x = x
        self.y = y

    def to_json(self) -> str:
        return json.dumps({"x": self.x, "y": self.y})


print(Point(1, 2).to_json())
````

### Challenge 36: Implement Hero Roles

Roles can be modeled as small strategy objects or as subclasses—both appear in design discussions.

### Challenge

Implement `Role` ABC with `title()`. `Tank` and `DPS` return different titles; `Hero` stores `Role` and prints `role.title()`.

````python
from abc import ABC, abstractmethod


class Role(ABC):
    # TODO
    pass


class Tank(Role):
    def title(self) -> str:
        return "Tank"


class DPS(Role):
    def title(self) -> str:
        return "DPS"


class Hero:
    def __init__(self, role: Role) -> None:
        self.role = role


print(Hero(Tank()).role.title())
````
