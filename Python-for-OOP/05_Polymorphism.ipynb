```python
class Shape:
    def name(self) -> str:
        raise NotImplementedError


class Circle(Shape):
    def name(self) -> str:
        return "circle"


class Square(Shape):
    def name(self) -> str:
        return "square"


print(", ".join(s.name() for s in [Circle(), Square()]))
```

```python
class Unit:
    def __init__(self, hp: int) -> None:
        self.hp = hp

    def act(self) -> None:
        raise NotImplementedError


class Fighter(Unit):
    def act(self) -> None:
        print("hit")


class Healer(Unit):
    def act(self) -> None:
        self.hp = min(50, self.hp + 5)


f = Fighter(10)
h = Healer(48)
f.act()
h.act()
print(h.hp)
```

<summary><strong>Show Solution</strong></summary>

```python
def area(shape: dict) -> float:
    kind = shape["kind"]
    if kind == "rect":
        return float(shape["w"] * shape["h"])
    if kind == "circle":
        return 3.14 * float(shape["r"]) ** 2
    raise ValueError("unknown kind")


print(area({"kind": "rect", "w": 3, "h": 4}))
print(area({"kind": "circle", "r": 2}))
```
