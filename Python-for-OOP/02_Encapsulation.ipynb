# Section 2: Encapsulation
*Challenges 9–15*

**Encapsulation** means hiding internal details and exposing a small, stable surface area. In Python, privacy is a convention: a leading underscore suggests “internal use”, while name mangling with `__` makes attributes harder to accidentally access from subclasses. For interviews, be ready to discuss getters, setters, and `@property` as ways to guard invariants.

### Challenge 9: Public Attribute and Method

Everything public by default is powerful, but it also means callers can put an object into a bad state unless you document expectations clearly.

````python
class Counter:
    def __init__(self) -> None:
        self.value = 0

    # TODO: inc/dec


x = Counter()
x.inc()
x.inc()
x.dec()
x.dec()
x.dec()
print(x.value)
````

<summary><strong>Show Solution</strong></summary>

```python
class Counter:
    def __init__(self) -> None:
        self.value = 0

    def inc(self) -> None:
        self.value += 1

    def dec(self) -> None:
        self.value = max(0, self.value - 1)


x = Counter()

x.inc()
x.dec()
x.dec()
x.dec()
```

```python
    def __init__(self, starting: float = 0.0) -> None:
        self._balance = float(starting)

    def _validate_amount(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError("amount must be positive")

    def deposit(self, amount: float) -> None:
        self._validate_amount(amount)
        self._balance += amount

    def withdraw(self, amount: float) -> None:
        self._validate_amount(amount)
        self._balance = max(0.0, self._balance - amount)

w.withdraw(7)
```

<details>
<summary><strong>Show Solution</strong></summary>

```python
class Rectangle:
    def __init__(self, w: float, h: float) -> None:
        if w <= 0 or h <= 0:
            raise ValueError("dimensions must be positive")
        self.__w = w
        self.__h = h

    def get_area(self) -> float:
        return self.__w * self.__h


r = Rectangle(2, 5)
print(r.get_area())
```

### Challenge 13: Property and Setter Decorator

`@property` lets you expose an attribute-like access pattern while running code. `@name.setter` validates writes.

```python
class Temperature:
    def __init__(self, celsius: float) -> None:
        self._c = float(celsius)

    @property
    def c(self) -> float:
        return self._c

    @property
    def f(self) -> float:
        return self._c * 9 / 5 + 32

    @f.setter
    def f(self, value: float) -> None:
```

### Challenge 15: Hide Details Using Decorators

Composition plus properties can enforce invariants even when requirements evolve.

### Challenge

Add `@property` for `email` returning lowercase; setter rejects emails without `@`.

````python
class User:
    def __init__(self, email: str) -> None:
        self._email = email.lower()

    @property
    def email(self) -> str:
        return self._email

    # TODO: email.setter validation


u = User("A@B.COM")
print(u.email)
````

<details>
<summary><strong>Show Solution</strong></summary>

```python
class User:
    def __init__(self, email: str) -> None:
        if "@" not in email:
            raise ValueError("invalid email")
        self._email = email.lower()

    @property
    def email(self) -> str:
        return self._email

    @email.setter
    def email(self, value: str) -> None:
        if "@" not in value:
            raise ValueError("invalid email")
        self._email = value.lower()


u = User("A@B.COM")
print(u.email)
```
