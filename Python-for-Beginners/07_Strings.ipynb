### Challenge 48: String Indexing

Each character in a string has an **index** (position). In Python, indexing starts at **0**.

```python
word = "Python"
# Index: 0  1  2  3  4  5
#        P  y  t  h  o  n

print(word[0])  # P
print(word[1])  # y
print(word[5])  # n
```

You can also use **negative indexing** to count from the end:

```python
print(word[-1])  # n (last character)
print(word[-2])  # o (second to last)
```

Trying to access an index that doesn't exist will raise an `IndexError`:

```python
print(word[10])  # IndexError!
```

### Challenge 49: String Looping

You can loop through each character in a string using its **index** with a `while` or `for` loop:

```python
word = "Hello"
i = 0
while i < len(word):
    print(word[i])
    i += 1
```

Output:
```
H
e
l
l
o
```

Using a `for` loop with `range()` and `len()`:

```python
word = "Hello"
for i in range(len(word)):
    print(f"Index {i}: {word[i]}")
```

This approach is useful when you need the **index** as well as the character.

### Challenge 50: String Looping Shorthand

Python provides a simpler way to loop through a string — you can iterate **directly** over the characters:

```python
word = "Hello"
for char in word:
    print(char)
```

Output:
```
H
e
l
l
o
```

This is cleaner and more "Pythonic" than using `range(len(word))`. Use this when you **don't need the index**, just the character.

If you need both the index and character, use `enumerate()`:

```python
for i, char in enumerate("Hello"):
    print(f"{i}: {char}")
```

### Challenge 51: String Concatenation

**String concatenation** is the process of joining two or more strings together using the `+` operator:

```python
first = "Hello"
second = "World"
result = first + " " + second
print(result)  # Hello World
```

You can also use the `*` operator to repeat a string:

```python
print("Ha" * 3)   # HaHaHa
print("-" * 20)   # --------------------
```

Remember, you can't concatenate a string with a number directly — you need to convert the number to a string first:

```python
age = 25
# print("Age: " + age)        # TypeError!
print("Age: " + str(age))      # Age: 25
```

### Challenge 52: String Slicing Part 1

**String slicing** lets you extract a portion of a string using the syntax `string[start:stop]`:

```python
text = "Hello, World!"
print(text[0:5])   # Hello
print(text[7:12])  # World
```

The slice starts at `start` (inclusive) and ends at `stop` (exclusive).

If you omit `start`, it defaults to the beginning:

```python
print(text[:5])    # Hello
```

If you omit `stop`, it goes to the end:

```python
print(text[7:])    # World!
```

If you omit both, you get a **copy** of the whole string:

```python
print(text[:])     # Hello, World!
```

### Challenge 53: String Slicing Part 2

String slicing also supports a **step** parameter: `string[start:stop:step]`.

```python
text = "0123456789"
print(text[0:10:2])  # 02468 (every 2nd character)
print(text[1:10:2])  # 13579 (every 2nd, starting at 1)
print(text[::3])     # 0369  (every 3rd character)
```

You can also use **negative indices** with slicing:

```python
text = "Hello, World!"
print(text[-6:])     # orld!
print(text[-6:-1])   # orld
```

### Challenge 54: Reversing a String

One of the most elegant tricks in Python is reversing a string using slicing with a step of `-1`:

```python
text = "Hello"
reversed_text = text[::-1]
print(reversed_text)  # olleH
```

The slice `[::-1]` means: start at the end, go to the beginning, step backwards by 1.

This is a very common pattern in Python and is frequently used in coding interviews to check if a string is a **palindrome** (reads the same forwards and backwards):

```python
word = "racecar"
if word == word[::-1]:
    print("Palindrome!")
else:
    print("Not a palindrome")
# Output: Palindrome!
```

### Challenge 55: Strings are Immutable

In Python, strings are **immutable** — once created, they cannot be changed. You cannot modify individual characters:

```python
word = "Hello"
word[0] = "J"  # TypeError! Strings don't support item assignment
```

Instead, you create a **new string**:

```python
word = "Hello"
new_word = "J" + word[1:]
print(new_word)  # Jello
```

Python provides several string methods that return **new strings**:

```python
text = "Hello, World!"
print(text.upper())       # HELLO, WORLD!
print(text.lower())       # hello, world!
print(text.replace("World", "Python"))  # Hello, Python!
print(text)               # Hello, World! (original unchanged!)
```

### Challenge 56: Strings Formatting

**f-strings** (formatted string literals) are the modern way to embed variables and expressions inside strings. They were introduced in Python 3.6.

```python
name = "Alice"
age = 25
print(f"My name is {name} and I am {age} years old.")
# Output: My name is Alice and I am 25 years old.
```

Just put an `f` before the opening quote, and use `{}` to embed any Python expression:

```python
x = 10
y = 3
print(f"{x} + {y} = {x + y}")    # 10 + 3 = 13
print(f"{x} / {y} = {x / y:.2f}") # 10 / 3 = 3.33
```

The `:.2f` is a **format specifier** that rounds to 2 decimal places.

f-strings are much cleaner than string concatenation:

```python
# Old way
print("Hello, " + name + "! You are " + str(age) + " years old.")

# New way (f-string)
print(f"Hello, {name}! You are {age} years old.")
```
