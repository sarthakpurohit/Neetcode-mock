# Section 5: Conditional Statements

Conditional statements allow your program to make decisions. Instead of running every line of code from top to bottom, you can control which code runs based on certain conditions. This is one of the most powerful concepts in programming.

### Challenge 31: Comparison Operators

**Comparison operators** compare two values and return a boolean (`True` or `False`). They are the foundation of conditional logic.

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater than or equal to | `5 >= 5` | `True` |
| `<=` | Less than or equal to | `3 <= 5` | `True` |

**Important:** Don't confuse `=` (assignment) with `==` (comparison).

```python
x = 10     # Assignment: set x to 10
x == 10    # Comparison: is x equal to 10? (True)
```

You can also compare strings:

```python
print("apple" == "apple")  # True
print("apple" == "Apple")  # False (case-sensitive)
print("apple" < "banana")  # True (alphabetical order)
```

### Challenge 32: If Statements

An **if statement** lets you execute code only when a condition is `True`.

```python
age = 18

if age >= 18:
    print("You are an adult.")
```

If the condition `age >= 18` is `True`, the indented code runs. If it's `False`, the indented code is skipped.

The **indentation** (4 spaces) is critical in Python. It tells Python which lines belong to the `if` block:

```python
if True:
    print("This is inside the if block")
    print("This is also inside")
print("This is outside the if block — always runs")
```

### Challenge 33: If Statement Scope

Unlike functions, `if` statements in Python **do not create a new scope**. Variables created inside an `if` block are accessible outside of it:

```python
x = 10

if x > 5:
    message = "x is big"

print(message)  # "x is big" — this works!
```

However, this can be risky. If the condition is `False`, the variable is never created:

```python
x = 3

if x > 5:
    message = "x is big"

print(message)  # NameError! message was never created
```

To avoid this, it's a good practice to **initialize variables before the if statement**:

```python
x = 3
message = "default"

if x > 5:
    message = "x is big"

print(message)  # "default" — safe!
```

### Challenge 34: If-Else Statements

An **if-else** statement lets you define what happens when the condition is `False`:

```python
age = 15

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

Output: `You are a minor.`

The `else` block runs **only if** the `if` condition is `False`. You can think of it as: "if this condition is true, do X; otherwise, do Y."

### Challenge 35: Else-If Statements

When you have **more than two conditions** to check, you can use `elif` (short for "else if"):

```python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(grade)  # B
```

Python checks each condition **in order**. As soon as one condition is `True`, its block runs and the rest are skipped.

You can have as many `elif` blocks as you want, and the `else` block is optional.

### Challenge 36: Logic Condition

You can combine multiple conditions using `and`, `or`, and `not` inside `if` statements:

```python
age = 25
has_license = True

if age >= 16 and has_license:
    print("You can drive!")
```

More examples:

```python
# OR: at least one condition must be True
if age < 12 or age > 65:
    print("Discount available!")

# NOT: flip the condition
if not has_license:
    print("You need a license.")

# Complex conditions with parentheses
if (age >= 18 and has_license) or age >= 21:
    print("Access granted")
```

### Challenge 37: Truthy and Falsy

In Python, every value has a **truthiness** — it's either considered `True` or `False` when used in a boolean context (like an `if` statement).

**Falsy values** (evaluate to `False`):
- `False`
- `0` (zero)
- `0.0` (zero float)
- `""` (empty string)
- `None`
- `[]` (empty list)
- `{}` (empty dict)

**Everything else is truthy** (evaluates to `True`):
- Non-zero numbers: `1`, `-5`, `3.14`
- Non-empty strings: `"hello"`, `" "`
- Non-empty collections: `[1, 2]`, `{"a": 1}`

```python
if "hello":
    print("This runs!")  # Non-empty string is truthy

if "":
    print("This won't run")  # Empty string is falsy

if 0:
    print("This won't run")  # 0 is falsy

if 42:
    print("This runs!")  # Non-zero is truthy
```
