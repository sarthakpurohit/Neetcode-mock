# Section 10: Dictionaries

Dictionaries are Python's key-value data structure. They let you store and retrieve data using meaningful keys instead of numeric indices. Dictionaries are one of the most used data structures in real-world Python programming.

### Challenge 69: Intro to Dictionaries

A **dictionary** (or `dict`) stores data as **key-value pairs**. Each key maps to a value.

```python
person = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}
```

Access values using their **key**:

```python
print(person["name"])  # Alice
print(person["age"])   # 25
```

You can also use `.get()`, which returns `None` (or a default) if the key doesn't exist:

```python
print(person.get("name"))      # Alice
print(person.get("email"))     # None
print(person.get("email", "N/A"))  # N/A
```

### Challenge 70: Dict Operations

Dictionaries are mutable — you can add, modify, and check for keys:

**Adding/modifying values:**

```python
person = {"name": "Alice"}
person["age"] = 25         # Add new key
person["name"] = "Bob"     # Modify existing key
print(person)  # {'name': 'Bob', 'age': 25}
```

**Checking if a key exists:**

```python
print("name" in person)   # True
print("email" in person)  # False
```

**Getting the number of key-value pairs:**

```python
print(len(person))  # 2
```

### Challenge 71: Dict Looping

You can loop through dictionaries in several ways:

**Loop through keys** (default):

```python
person = {"name": "Alice", "age": 25}

for key in person:
    print(key)
# name
# age
```

**Loop through values:**

```python
for value in person.values():
    print(value)
# Alice
# 25
```

**Loop through key-value pairs:**

```python
for key, value in person.items():
    print(f"{key}: {value}")
# name: Alice
# age: 25
```

### Challenge 72: Dict Practice

Dictionaries are perfect for **counting occurrences** of items:

```python

for char in text:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1

print(char_count)  # {'h': 1, 'e': 1, 'l': 2, 'o': 1}
```

### Challenge 73: Dict Remove

There are several ways to remove items from a dictionary:

**`del`** — removes a key-value pair:

```python
person = {"name": "Alice", "age": 25, "city": "NYC"}
del person["age"]
print(person)  # {'name': 'Alice', 'city': 'NYC'}
```

**`.pop()`** — removes and returns the value:

```python
city = person.pop("city")
print(city)    # NYC
print(person)  # {'name': 'Alice'}
```

**`.pop()` with default** — avoids `KeyError`:

```python
email = person.pop("email", "not found")
print(email)  # not found
```

**`.clear()`** — removes all items:

```python
person.clear()
print(person)  # {}
```

### Challenge 74: Dict Values

Dictionaries provide several useful methods for accessing their contents:

```python
person = {"name": "Alice", "age": 25, "city": "NYC"}

print(person.keys())    # dict_keys(['name', 'age', 'city'])
print(person.values())  # dict_values(['Alice', 25, 'NYC'])
print(person.items())   # dict_items([('name', 'Alice'), ('age', 25), ('city', 'NYC')])
```

You can convert these to regular lists:

```python
keys_list = list(person.keys())
print(keys_list)  # ['name', 'age', 'city']
```
