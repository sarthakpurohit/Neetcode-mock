Variables are one of the most fundamental concepts in programming. They allow you to store and reuse values throughout your code. In this section, you'll learn how to declare variables, follow naming conventions, understand types, and work with Python's dynamic typing system.

### Challenge 6: Variable Declaration

Earlier we printed the "Hello, world!" string to the console. After it was printed we couldn't reuse that string, unless we retyped it all out from scratch, as shown below:

```python
print("Hello, world!")
print("Hello, world!")
```

But what if we wanted to use that string multiple times in our code? This is where **variables** come in. With the following code, we can store the string in a variable and print it multiple times:

```python
message = "Hello, world!"
print(message)
print(message)
```

Output:

```
Hello, world!
Hello, world!
```

In the code above, we stored the string `"Hello, world!"` in a variable called `message`. Then, instead of passing a raw string to the `print()` function, we passed the variable `message`. This way, we can print the same string multiple times without having to retype it.

Variables are like containers that hold values.

### Challenge 7: Variable Naming

In Python, variable names must follow certain rules. If you break these rules, Python will give you a `SyntaxError`.

Here are the rules for naming variables in Python:

1. Variable names can only contain **letters** (a-z, A-Z), **digits** (0-9), and **underscores** (`_`).
2. Variable names **cannot start with a digit**.
3. Variable names are **case-sensitive** — `myVar` and `myvar` are different variables.
4. Variable names **cannot be Python reserved words** (keywords) like `if`, `for`, `while`, `class`, `return`, etc.

```python
# Valid variable names
my_name = "Alice"
age2 = 25
_private = True

# Invalid variable names
# 2nd_place = "Bob"   # starts with a digit
# my-name = "Alice"   # contains a hyphen
# class = "Math"      # reserved word
```

### Challenge 8: Naming Conventions

While Python allows many valid variable names, the Python community has agreed on certain **naming conventions** to make code more readable.

The most common convention in Python is **snake_case**, where words are separated by underscores and all letters are lowercase:

```python
my_variable = 10
first_name = "Alice"
total_count = 42
```

Other languages use different conventions:
- **camelCase**: `myVariable`, `firstName` — common in JavaScript and Java
- **PascalCase**: `MyVariable`, `FirstName` — used for class names in Python and many other languages
- **UPPER_SNAKE_CASE**: `MAX_SIZE`, `PI_VALUE` — used for constants in Python

In Python, `snake_case` is the standard for variable and function names. `PascalCase` is used for class names, and `UPPER_SNAKE_CASE` is used for constants.

### Challenge 9: Reassigning Variables

In Python, variables can be **reassigned** to new values at any time. When you assign a new value to an existing variable, the old value is replaced.

```python
x = 10
print(x)  # Output: 10

x = 20
print(x)  # Output: 20
```

The variable `x` first holds the value `10`, and after reassignment, it holds the value `20`. The old value `10` is discarded.

You can even reassign a variable to a completely different type of value:

```python
x = 10       # x is an integer
x = "hello"  # x is now a string
```

This is possible because Python is **dynamically typed**, which we'll learn more about later.

### Challenge 10: Multiple Assignments

Python allows you to assign values to multiple variables in a single line. This is called **multiple assignment** or **tuple unpacking**.

```python
a, b, c = 1, 2, 3
print(a)  # Output: 1
print(b)  # Output: 2
print(c)  # Output: 3
```

You can also assign the **same value** to multiple variables at once:

```python
x = y = z = 0
print(x)  # Output: 0
print(y)  # Output: 0
print(z)  # Output: 0
```

Multiple assignment is a convenient feature that makes your code more concise. It's commonly used when you want to initialize several variables at once, or when swapping values:

```python
a, b = 1, 2
a, b = b, a  # Swap!
print(a)  # Output: 2
print(b)  # Output: 1
```

### Challenge 11: Variable Types

Every value in Python has a **type**. The type tells Python what kind of data the value represents and what operations can be performed on it.

The most common built-in types are:

| Type | Description | Example |
|------|-------------|--------|
| `int` | Integer (whole number) | `42` |
| `float` | Decimal number | `3.14` |
| `str` | String (text) | `"hello"` |
| `bool` | Boolean (True/False) | `True` |

You can check the type of any value using the `type()` function:

```python
print(type(42))       # <class 'int'>
print(type(3.14))     # <class 'float'>
print(type("hello"))  # <class 'str'>
print(type(True))     # <class 'bool'>
```

### Challenge 12: Dynamic Typing

Python is a **dynamically typed** language. This means you don't need to declare the type of a variable when you create it — Python figures out the type automatically based on the value you assign.

```python
x = 10        # x is an int
x = "hello"   # x is now a str
x = 3.14      # x is now a float
```

This is different from **statically typed** languages like Java or C++, where you must declare the type of a variable when you create it, and you cannot change it later:

```java
// Java example (statically typed)
int x = 10;
x = "hello";  // Error! x must be an int
```

Dynamic typing makes Python flexible and easy to use, especially for beginners. However, it can also lead to bugs if you accidentally assign the wrong type of value to a variable.

### Challenge 13: Type Casting

Sometimes you need to convert a value from one type to another. This is called **type casting** (or type conversion).

Python provides built-in functions for type casting:

- `int()` — converts to integer
- `float()` — converts to float
- `str()` — converts to string
- `bool()` — converts to boolean

```python
# String to integer
age = int("25")
print(age)        # 25
print(type(age))  # <class 'int'>

# Integer to string
num_str = str(100)
print(num_str)        # 100
print(type(num_str))  # <class 'str'>

# Integer to float
pi = float(3)
print(pi)  # 3.0
```

Be careful — not all conversions are valid:

```python
int("hello")  # ValueError! Can't convert "hello" to int
```

### Challenge 14: Type Errors

A **TypeError** occurs when you try to perform an operation on values of incompatible types. For example, you cannot add a string and an integer together:

```python
result = "age: " + 25  # TypeError!
```

Python doesn't know how to combine a string and an integer with `+`. Should it convert the number to a string? Or the string to a number? Python doesn't guess — it raises a `TypeError` instead.

To fix this, you need to explicitly convert one of the values:

```python
result = "age: " + str(25)  # "age: 25"
```

Or you can use an f-string (which we'll learn about later):

```python
result = f"age: {25}"  # "age: 25"
```

### Challenge 15: Empty Variable

In Python, there is a special value called `None` that represents the **absence of a value**. It is Python's way of saying "nothing" or "no value here".

```python
x = None
print(x)       # None
print(type(x)) # <class 'NoneType'>
```

`None` is commonly used to:
- Initialize a variable before assigning it a real value
- Represent a missing or default value
- Indicate that a function doesn't return anything

```python
result = None

if True:
    result = "found"

print(result)  # "found"
```

You can check if a variable is `None` using the `is` keyword:

```python
x = None
if x is None:
    print("x has no value")
```
