# Section 12: Exception Handling

Errors are inevitable in programming. Exception handling lets you anticipate and gracefully deal with errors instead of letting your program crash. In this section, you'll learn how to use `try`, `except`, and handle specific error types.

### Challenge 79: Try Except

The `try`/`except` block lets you handle errors gracefully. Code that might cause an error goes in the `try` block, and the error-handling code goes in the `except` block.

```python
try:
    result = 10 / 0
except:
    print("An error occurred!")
```

Output: `An error occurred!`

Without `try`/`except`, dividing by zero would crash the program with a `ZeroDivisionError`.

The code in `except` only runs if an error occurs. If no error occurs, the `except` block is skipped:

```python
try:
    result = 10 / 2
    print(result)  # 5.0
except:
    print("An error occurred!")  # This doesn't run
```

### Challenge 80: Error Catching

It's best practice to catch **specific** exception types rather than using a bare `except`. This helps you handle different errors differently.

```python
try:
    num = int("abc")
except ValueError:
    print("That's not a valid number!")
```

Common exception types:

| Exception | When it occurs |
|-----------|---------------|
| `ValueError` | Invalid value conversion |
| `TypeError` | Wrong type operation |
| `ZeroDivisionError` | Division by zero |
| `IndexError` | List index out of range |
| `KeyError` | Dictionary key not found |
| `NameError` | Undefined variable |
| `FileNotFoundError` | File doesn't exist |

You can also access the error message:

```python
try:
    num = int("abc")
except ValueError as e:
    print(f"Error: {e}")
# Error: invalid literal for int() with base 10: 'abc'
```

### Challenge 81: Multiple Except Blocks

You can handle **multiple exception types** with separate `except` blocks:

```python
try:
    numbers = [1, 2, 3]
    print(numbers[10])     # IndexError
    print(10 / 0)          # ZeroDivisionError
except IndexError:
    print("Index out of range!")
except ZeroDivisionError:
    print("Cannot divide by zero!")
```

Only the **first** error that occurs gets caught — once an error happens, the rest of the `try` block is skipped.

You can also add `else` and `finally`:

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Error!")
else:
    print(f"Result: {result}")  # Runs only if no error
finally:
    print("Done!")  # Always runs, error or not
```
