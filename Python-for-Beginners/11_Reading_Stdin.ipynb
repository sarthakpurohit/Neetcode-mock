### Challenge 75: Reading Input

The `input()` function reads a line of text from the user (standard input, or "stdin").

```python
name = input("What is your name? ")
print("Hello, " + name + "!")
```

When this runs, the program pauses and waits for the user to type something and press Enter. The text they type is returned as a **string**.

You can also call `input()` without a prompt:

```python
data = input()
print("You entered: " + data)
```

**Important:** `input()` **always returns a string**, even if the user types a number.

```python
age = input("Age: ")  # User types 25
print(type(age))       # <class 'str'>
```

### Challenge 76: Type Conversion with Input

Since `input()` always returns a string, you often need to **convert** the input to the correct type.

```python
# Reading an integer
age = int(input("Enter your age: "))
print(age + 1)  # Math works because age is now an int

# Reading a float
price = float(input("Enter the price: "))
print(f"Tax: {price * 0.08:.2f}")
```

**Be careful!** If the user enters something that can't be converted, you'll get a `ValueError`:

```python
age = int(input())  # User types "abc"
# ValueError: invalid literal for int() with base 10: 'abc'
```

We'll learn how to handle this with exception handling later.

### Challenge 77: Parse Input

Sometimes you need to read multiple values from a single line of input. The `.split()` method is perfect for this:

```python
line = input()  # User types: "Alice 25"
parts = line.split()  # ['Alice', '25']
name = parts[0]
age = int(parts[1])
print(f"{name} is {age} years old")
```

`.split()` splits a string by whitespace (spaces) by default. You can also split by a specific character:

```python
line = "apple,banana,cherry"
items = line.split(",")
print(items)  # ['apple', 'banana', 'cherry']
```

You can combine splitting with **unpacking**:

```python
name, age = input().split()
age = int(age)
```

### Challenge 78: Read Input Practice

Let's combine everything we've learned about input to build a simple calculator.

A common pattern is reading input in a loop until a certain condition is met:

```python
while True:
    command = input("Enter command (or 'quit'): ")
    if command == "quit":
        break
    print(f"You entered: {command}")
```

You can also read a number of inputs specified by the first line:

```python
n = int(input())  # Number of items
for i in range(n):
    item = input()
    print(f"Item {i + 1}: {item}")
```
