### Challenge 16: Arithmetic Operators

Python supports all the basic arithmetic operators you're familiar with from math class, plus a few extras:

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division | `7 / 2` | `3.5` |
| `//` | Floor Division | `7 // 2` | `3` |
| `%` | Modulo (Remainder) | `7 % 2` | `1` |
| `**` | Exponentiation | `2 ** 3` | `8` |

A few things to note:
- `/` always returns a `float`, even if the result is a whole number: `6 / 3` gives `2.0`
- `//` performs **floor division** — it divides and rounds down to the nearest integer
- `%` gives the **remainder** after division — very useful for checking if a number is even or odd
- `**` is for **exponentiation** (raising to a power)

```python
print(10 + 3)   # 13
print(10 - 3)   # 7
print(10 * 3)   # 30
print(10 / 3)   # 3.3333...
print(10 // 3)  # 3
print(10 % 3)   # 1
print(10 ** 3)  # 1000
```

### Challenge 17: More Operators

Python follows the standard **order of operations** (PEMDAS/BODMAS) when evaluating mathematical expressions:

1. **P**arentheses `()`
2. **E**xponentiation `**`
3. **M**ultiplication `*`, **D**ivision `/`, Floor Division `//`, Modulo `%`
4. **A**ddition `+`, **S**ubtraction `-`

When operators have the same precedence, they are evaluated **left to right**.

```python
result = 2 + 3 * 4      # 14, not 20 (multiplication first)
result = (2 + 3) * 4    # 20 (parentheses first)
result = 2 ** 3 ** 2     # 512 (exponentiation is right-to-left!)
```

Parentheses are your best friend — use them to make your intent clear, even when they aren't strictly necessary.

You can also perform arithmetic with variables:

```python
price = 19.99
tax_rate = 0.08
total = price + (price * tax_rate)
print(total)  # 21.5892
```

### Challenge 18: Shorthand Operators

Python provides **shorthand** (or **augmented assignment**) operators that let you update a variable's value more concisely.

Instead of writing `x = x + 5`, you can write `x += 5`. Here are the shorthand operators:

| Shorthand | Equivalent | Example |
|-----------|-----------|--------|
| `x += 5` | `x = x + 5` | Addition |
| `x -= 5` | `x = x - 5` | Subtraction |
| `x *= 5` | `x = x * 5` | Multiplication |
| `x /= 5` | `x = x / 5` | Division |
| `x //= 5` | `x = x // 5` | Floor Division |
| `x %= 5` | `x = x % 5` | Modulo |
| `x **= 5` | `x = x ** 5` | Exponentiation |

```python
score = 100
score += 10   # score is now 110
score -= 20   # score is now 90
score *= 2    # score is now 180
print(score)  # 180
```

### Challenge 19: Boolean OR

A **boolean** is a value that is either `True` or `False`. Booleans are named after the mathematician George Boole, who developed a system of logic based on these two values.

The `or` operator returns `True` if **at least one** of the operands is `True`:

| A | B | A or B |
|---|---|--------|
| `True` | `True` | `True` |
| `True` | `False` | `True` |
| `False` | `True` | `True` |
| `False` | `False` | `False` |

```python
print(True or True)    # True
print(True or False)   # True
print(False or True)   # True
print(False or False)  # False
```

The `or` operator is commonly used in conditions:

```python
is_weekend = True
is_holiday = False
can_sleep_in = is_weekend or is_holiday
print(can_sleep_in)  # True
```

### Challenge 20: Boolean AND

The `and` operator returns `True` only if **both** operands are `True`:

| A | B | A and B |
|---|---|--------|
| `True` | `True` | `True` |
| `True` | `False` | `False` |
| `False` | `True` | `False` |
| `False` | `False` | `False` |

```python
print(True and True)    # True
print(True and False)   # False
print(False and True)   # False
print(False and False)  # False
```

The `and` operator is commonly used when you need **multiple conditions** to be true:

```python
has_ticket = True
has_id = True
can_enter = has_ticket and has_id
print(can_enter)  # True
```

### Challenge 21: Boolean Negation

The `not` operator **flips** a boolean value — `True` becomes `False`, and `False` becomes `True`.

```python
print(not True)   # False
print(not False)  # True
```

You can combine `not` with `and` and `or` to create more complex conditions:

```python
is_raining = False
has_umbrella = True

should_go_outside = not is_raining or has_umbrella
print(should_go_outside)  # True
```

**Operator precedence** for boolean operators (highest to lowest):
1. `not`
2. `and`
3. `or`

```python
print(not True or False)      # False or False = False
print(not (True or False))    # not True = False
print(True or not False)      # True or True = True
```
