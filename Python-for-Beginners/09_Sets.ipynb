# Section 9: Sets

Sets are unordered collections of **unique** elements. They are incredibly useful for removing duplicates, testing membership, and performing mathematical set operations like union and intersection.

### Challenge 66: Intro to Sets

A **set** is an unordered collection of **unique items**. Duplicate values are automatically removed.

```python
fruits = {"apple", "banana", "cherry", "apple"}
print(fruits)  # {'apple', 'banana', 'cherry'}
```

Sets are created with curly braces `{}` or the `set()` function:

```python
empty_set = set()  # Note: {} creates an empty dictionary, not a set!
numbers = {1, 2, 3}
from_list = set([1, 2, 2, 3, 3, 3])
print(from_list)  # {1, 2, 3}
```

Key characteristics of sets:
- **Unordered** — items have no index
- **Unique** — no duplicate elements
- **Mutable** — you can add/remove items
- **Fast membership testing** — `in` is very fast with sets

```python
numbers = {1, 2, 3, 4, 5}
print(3 in numbers)   # True
print(10 in numbers)  # False
print(len(numbers))   # 5
```

### Challenge 67: Set Operations

Sets support mathematical set operations:

**Union** (`|` or `.union()`) — all items from both sets:

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a | b)  # {1, 2, 3, 4, 5}
```

**Intersection** (`&` or `.intersection()`) — items in both sets:

```python
print(a & b)  # {3}
```

**Difference** (`-` or `.difference()`) — items in first but not second:

```python
print(a - b)  # {1, 2}
print(b - a)  # {4, 5}
```

**Symmetric Difference** (`^` or `.symmetric_difference()`) — items in either set but not both:

```python
print(a ^ b)  # {1, 2, 4, 5}
```

### Challenge 68: Set Practice

Let's practice using sets to solve a common problem: finding unique elements.

You can **add** items with `.add()` and **remove** items with `.remove()` or `.discard()`:

```python
colors = {"red", "green"}
colors.add("blue")
print(colors)  # {'red', 'green', 'blue'}

colors.remove("red")
print(colors)  # {'green', 'blue'}
```

`.remove()` raises a `KeyError` if the item doesn't exist. `.discard()` does nothing:

```python
colors.discard("purple")  # No error, even though 'purple' isn't in the set
```
