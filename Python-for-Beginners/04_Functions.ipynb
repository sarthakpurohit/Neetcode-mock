# Section 4: Functions

Functions are reusable blocks of code that perform a specific task. They are one of the most important concepts in programming. In this section, you'll learn how to define functions, use parameters, return values, and understand variable scope.

### Challenge 22: Introduction to Functions

A **function** is a reusable block of code that performs a specific task. Functions help you organize your code and avoid repeating the same code over and over.

You've already been using a function: `print()`. It's a **built-in function** provided by Python. But you can also create your own functions!

Here's a simple function:

```python
def greet():
    print("Hello!")
    print("Welcome to Python!")
```

Let's break this down:
- `def` is the keyword that tells Python you're **defining** a function
- `greet` is the **name** of the function
- `()` is where **parameters** go (empty for now)
- `:` marks the start of the function **body**
- The **indented** lines are the code that runs when the function is called

To **call** (use) the function, write its name followed by parentheses:

```python
greet()  # Calls the function
```

### Challenge 23: Function Declaration

When you define a function, the code inside it **does not run** until you call the function. This is an important distinction:

```python
def say_hi():
    print("Hi!")

print("Before calling the function")
say_hi()
print("After calling the function")
```

Output:
```
Before calling the function
Hi!
After calling the function
```

You can call a function **as many times as you want**:

```python
say_hi()
say_hi()
say_hi()
```

This prints `Hi!` three times.

**Important:** You must define a function **before** you call it. Python reads code top to bottom, so if you try to call a function before defining it, you'll get a `NameError`.

### Challenge 24: Parameters

Functions become much more powerful when they accept **parameters** (also called **arguments**). Parameters are values you pass into the function so it can use them.

```python
def greet(name):
    print("Hello, " + name + "!")

greet("Alice")  # Output: Hello, Alice!
greet("Bob")    # Output: Hello, Bob!
```

In this example, `name` is a **parameter** — a placeholder for the value that will be passed in. When we call `greet("Alice")`, the string `"Alice"` is the **argument** — the actual value passed to the function.

Think of a parameter as a variable that only exists inside the function.

### Challenge 25: Multiple Parameters

Functions can accept **multiple parameters**, separated by commas:

```python
def add(a, b):
    print(a + b)

add(3, 5)   # Output: 8
add(10, 20) # Output: 30
```

The **order** of the arguments matters. The first argument is assigned to the first parameter, the second to the second, and so on:

```python
def introduce(name, age):
    print(name + " is " + str(age) + " years old.")

introduce("Alice", 25)  # Alice is 25 years old.
introduce(25, "Alice")  # 25 is Alice years old. (Wrong!)
```

### Challenge 26: Return Statement

So far, our functions have used `print()` to display results. But often you want a function to **send a value back** to the caller so you can use it in further computations. This is done with the `return` statement.

```python
def add(a, b):
    return a + b

result = add(3, 5)
print(result)  # 8
```

Key differences between `print()` and `return`:
- `print()` displays a value to the console but the function returns `None`
- `return` sends a value back to the caller, which can be stored in a variable

```python
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b

result1 = add_print(3, 5)  # Prints 8, but result1 is None
result2 = add_return(3, 5) # result2 is 8
```

Once `return` is executed, the function **immediately stops** — any code after `return` won't run.

### Challenge 27: Type Hints

Python is dynamically typed, meaning you don't have to declare what type a variable is. However, Python supports **type hints** (also called type annotations) that let you indicate what types you *expect*.

Type hints don't change how your code runs — they're just **documentation** for you and other developers.

```python
def add(a: int, b: int) -> int:
    return a + b
```

In this example:
- `a: int` means parameter `a` is expected to be an integer
- `b: int` means parameter `b` is expected to be an integer
- `-> int` means the function is expected to return an integer

You can also use type hints with variables:

```python
name: str = "Alice"
age: int = 25
is_student: bool = True
```

Type hints are widely used in professional Python code and help tools like IDEs give better autocomplete suggestions.

### Challenge 28: Scope

**Scope** refers to where in your code a variable can be accessed. Variables created inside a function have **local scope** — they can only be used inside that function.

```python
def my_function():
    x = 10  # Local variable
    print(x)

my_function()  # Output: 10
print(x)       # NameError! x is not defined outside the function
```

The variable `x` exists only inside `my_function()`. When the function finishes, `x` is destroyed.

Variables created outside of any function have **global scope** — they can be accessed anywhere in the code:

```python
y = 20  # Global variable

def my_function():
    print(y)  # Can access y here

my_function()  # Output: 20
print(y)       # Output: 20
```

### Challenge 29: Global vs Local Scope

What happens when a local variable and a global variable have the **same name**?

```python
x = "global"

def my_function():
    x = "local"
    print(x)

my_function()  # Output: local
print(x)       # Output: global
```

Inside the function, `x` refers to the **local** variable. Outside the function, `x` refers to the **global** variable. They are two separate variables that happen to share the same name.

If you want to **modify** the global variable from inside a function, you need to use the `global` keyword:

```python
x = "global"

def my_function():
    global x
    x = "modified"

my_function()
print(x)  # Output: modified
```

**Tip:** Using `global` is generally discouraged. It's better to pass values as parameters and use `return` to send them back.

### Challenge 30: Default Arguments

You can give function parameters **default values**. If the caller doesn't provide an argument, the default value is used instead.

```python
def greet(name="World"):
    print("Hello, " + name + "!")

greet("Alice")  # Output: Hello, Alice!
greet()          # Output: Hello, World!
```

In this example, if no argument is passed, `name` defaults to `"World"`.

You can have **some parameters with defaults and some without**. Parameters without defaults must come first:

```python
def order(item, quantity=1):
    print(str(quantity) + "x " + item)

order("Coffee")      # Output: 1x Coffee
order("Coffee", 3)   # Output: 3x Coffee
```
