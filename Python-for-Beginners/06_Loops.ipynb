# Section 6: Loops

Loops let you repeat a block of code multiple times. They are essential for processing data, iterating over collections, and automating repetitive tasks. In this section, you'll learn about `while` loops, `for` loops, and loop control flow.

### Challenge 38: While Loops

A **while loop** repeats a block of code **as long as a condition is `True`**.

```python
count = 0

while count < 3:
    print(count)
    count += 1
```

Output:
```
0
1
2
```

Here's how it works:
1. Check the condition: Is `count < 3`? Yes (`count` is 0)
2. Run the body: print `0`, then increment `count` to 1
3. Check the condition again: Is `count < 3`? Yes (`count` is 1)
4. Run the body: print `1`, then increment `count` to 2
5. Repeat until the condition is `False`

**Warning:** If the condition never becomes `False`, you'll get an **infinite loop** — your program will run forever!

### Challenge 39: While Loops Counting

A common use of while loops is **counting**. You can count up, count down, or count by any increment.

**Counting down:**

```python
n = 5
while n > 0:
    print(n)
    n -= 1
print("Liftoff!")
```

Output:
```
5
4
3
2
1
Liftoff!
```

**Counting by twos:**

```python
n = 0
while n <= 10:
    print(n)
    n += 2
# Output: 0, 2, 4, 6, 8, 10
```

### Challenge 40: While Loops Multiples

You can use while loops to generate **multiples** of a number. Remember, the modulo operator `%` is useful for checking divisibility.

```python
# Print multiples of 3 up to 15
n = 3
while n <= 15:
    print(n)
    n += 3
```

Output:
```
3
6
9
12
15
```

Another approach using modulo:

```python
n = 1
while n <= 15:
    if n % 3 == 0:
        print(n)
    n += 1
```

This also prints 3, 6, 9, 12, 15 but checks every number instead of jumping by 3.

### Challenge 41: For Loops

A **for loop** iterates over a sequence (like a range of numbers). Python's `range()` function generates a sequence of numbers:

```python
for i in range(5):
    print(i)
```

Output:
```
0
1
2
3
4
```

`range(5)` generates numbers from 0 to 4 (5 numbers total). Notice that it **starts at 0** and **stops before 5**.

This is the most common way to repeat something a specific number of times:

```python
for i in range(3):
    print("Hello!")
# Prints "Hello!" three times
```

### Challenge 43: For Loops Step

```python

    print(i)
```

Output:
```

2
4
6
8
```

The loop starts at 0, increments by 2 each time, and stops before 10.

You can use any step value:

```python

    print(i)
# Output: 0, 5, 10, 15, 20, 25
```

### Challenge 44: For Loops Reverse

You can count **backwards** with a **negative step**:

```python
for i in range(5, 0, -1):
    print(i)
```

Output:
```
5
4
3
2
1
```

Note that `range(5, 0, -1)` starts at 5 and goes down to 1 (stops before 0).

You can also use `reversed()` with `range()`:

```python
for i in reversed(range(1, 6)):
    print(i)
# Same output: 5, 4, 3, 2, 1
```

### Challenge 45: Nested Loops

You can put a loop **inside** another loop. This is called a **nested loop**. The inner loop runs completely for each iteration of the outer loop.

```python
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={j}")
```

Output:
```
i=0, j=0
i=0, j=1
i=1, j=0
i=1, j=1
i=2, j=0
i=2, j=1
```

For each value of `i`, the inner loop runs through all values of `j`. That's 3 × 2 = 6 total iterations.

Nested loops are commonly used with 2D data structures like grids or matrices.

### Challenge 46: Control Flow

Python provides two keywords to control the flow of loops:

**`break`** — exits the loop immediately:

```python
for i in range(10):
    if i == 5:
        break
    print(i)
# Output: 0, 1, 2, 3, 4
```

**`continue`** — skips the rest of the current iteration and moves to the next:

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
# Output: 0, 1, 3, 4
```

These work in both `for` and `while` loops.
